```python
import torch
import torch.nn as nn
import torch.optim as optim

# 1. Define the network
class SimpleNet(nn.Module):
    def __init__(self, input_size=10, hidden_size=32, output_size=10):
        super().__init__()
        self.layers = nn.Sequential(
            nn.Linear(input_size, hidden_size),
            nn.ReLU(),
            nn.Linear(hidden_size, output_size),
            nn.Sigmoid()
        )

    def forward(self, X):
        return self.layers(X)

# 2. Toy data (XOR problem)
X = torch.randn(4,10)
print(X)
y = torch.randn(4,10)
print(y)

# 3. Set up model, loss, optimizer
model = SimpleNet()
criterion = nn.MSELoss()
optimizer = optim.Adam(model.parameters(), lr=0.01)

torch.manual_seed(528)

# 4. Training loop
for epoch in range(1000):
    optimizer.zero_grad()
    outputs = model(X)
    loss = criterion(outputs, y)
    loss.backward()
    optimizer.step()

    if (epoch+ 1) % 200 == 0:
        print(f"Epoch {epoch+1}, Loss: {loss.item():.4f}")

# 5. Test
with torch.no_grad():
    predictions = model(X)
    print("Predictions:", predictions.squeeze().tolist())
```

```python
class neural(nn.Module):
  def __init__(self, ins = 10, hs=32, os=10):
    super().__init__()
    self.layers = nn.Sequential(
        nn.Linear(ins,hs),
        nn.ReLU(),
        nn.Linear(hs,os),
        nn.Sigmoid()
    )
  def forward(self, x):
    return self.layers(x)

x = torch.randn(4,10)
y = torch.randn(4,10)

model = neural()
criterion = nn.MSELoss()
optimizer= optim.Adam(model.parameters(), lr=0.01)

for epoch in range(1000):
  optimizer.zero_grad()
  outputs = model(x)
  loss = criterion(outputs, y)
  loss.backward()
  optimizer.step()

  if epoch %200 == 0:
    print(f"epoch: {epoch}, loss: {loss.item():.4f}")
```
